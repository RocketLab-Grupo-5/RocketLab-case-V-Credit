# Organização do ambiente

Prepara a estrutura do Unity Catalog para a arquitetura medalhão: o catálogo `vcredit_medallion`, os schemas `bronze`, `silver` e `gold` e o volume `landing`, onde ficam os CSVs de origem. É o primeiro notebook a ser executado, e pode ser executado de novo sem efeito, porque só cria o que ainda não existe.

## 1. Parâmetros

Nomes do catálogo, dos schemas e do caminho do volume usados por todos os notebooks.

```python
catalog ="vcredit_medallion"
bronze_schema_name = "bronze"
silver_schema_name = "silver"
gold_schema_name = "gold"

# criação dos nomes/caminhos dos schemas de cada camada
bronze_schema = f"{catalog}.{bronze_schema_name}"
silver_schema = f"{catalog}.{silver_schema_name}"
gold_schema = f"{catalog}.{gold_schema_name}"

# criação do nome/caminho volume para csvs
landing_path = f"/Volumes/{catalog}/{bronze_schema_name}/landing"

# verificação da configuração
print(f"catalog: {catalog}")
print(f"bronze_schema: {bronze_schema}")
print(f"silver_schema: {silver_schema}")
print(f"gold_schema: {gold_schema}")
print(f"landing_path: {landing_path}")
```

## 2. Criação da estrutura

Cria o catálogo, os três schemas e o volume `landing`, e exibe o resultado.

```python
# criação dos schemas de cada camada no banco
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {bronze_schema}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {silver_schema}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {gold_schema}")

display(spark.sql(f"SHOW SCHEMAS IN {catalog}"))

# criação do volume para csvs
spark.sql(f"CREATE VOLUME IF NOT EXISTS {bronze_schema}.landing")

display(spark.sql(f"SHOW VOLUMES IN {bronze_schema}"))
```
